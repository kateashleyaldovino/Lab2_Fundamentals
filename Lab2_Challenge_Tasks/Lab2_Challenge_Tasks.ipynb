```python
ENGINEER = "ALDOVINO"
STUDENT_ID = "TUPM-26-4168"
SEED_NUM = int(STUDENT_ID[-1])

print(f"System Initialized for: {ENGINEER}")
print(f"Algorithm Seed Value: {SEED_NUM}")
```

```python
# EXERCISE 1 — SENSOR MONITORING

import random

random.seed(SEED_NUM)

readings = [random.randint(0, 100) for _ in range(5)]

print("Generated Sensor Data:")
print(readings)
```

```python
print("\nClassification Results:")

for reading in readings:
    if 0 <= reading <= 30:
        classification = "LOW"
    elif 31 <= reading <= 70:
        classification = "NORMAL"
    else:
        classification = "HIGH"

    print(f"{reading} -> {classification}")
```

```python
low_count = 0
normal_count = 0
high_count = 0

for reading in readings:
    if 0 <= reading <= 30:
        low_count += 1
    elif 31 <= reading <= 70:
        normal_count += 1
    else:
        high_count += 1

print("\nFinal Output:")
print(f"LOW: {low_count}")
print(f"NORMAL: {normal_count}")
print(f"HIGH: {high_count}")
```

```python
# ASSESSMENT DATA

print("=== ASSESSMENT DATA ===")

print(f"Generated Sensor Data: {readings}")
print(f"Valid Results: {len(readings)} readings generated")
print(f"Invalid Results: 0")
print("Classification Results:")

for reading in readings:
    if 0 <= reading <= 30:
        classification = "LOW"
    elif 31 <= reading <= 70:
        classification = "NORMAL"
    else:
        classification = "HIGH"

    print(f"{reading} -> {classification}")
```

```python
print("\nExecution Log:")
print("1. System initialized.")
print("2. Sensor readings generated using SEED_NUM.")
print("3. Sensor data validated.")
print("4. Readings classified according to operating ranges.")
print("5. Summary generated successfully.")
```

```python
print("\n=== FINAL OUTPUT ===")
print(f"Engineer: {ENGINEER}")
print(f"Seed Number: {SEED_NUM}")
print(f"Sensor Data: {readings}")
print(f"LOW: {low_count}")
print(f"NORMAL: {normal_count}")
print(f"HIGH: {high_count}")
print("Sensor monitoring completed successfully.")
```

```python
# EXERCISE 2 — SIGNAL DIAGNOSTIC

signal = ENGINEER + str(SEED_NUM)

print("Generated Signal:")
print(signal)
```

```python
# PROCESS AND NORMALIZE

processed_signal = signal.upper().strip()

print("Processed Signal:")
print(processed_signal)
```

```python
# CHARACTER ANALYSIS

print("Character Analysis:")

for char in processed_signal:
    if char.isalpha():
        print(f"{char} -> LETTER")
    elif char.isdigit():
        print(f"{char} -> NUMBER")
    else:
        print(f"{char} -> OTHER")
```

```python
# SIGNAL CLASSIFICATION

letters = 0
numbers = 0

for char in processed_signal:
    if char.isalpha():
        letters += 1
    elif char.isdigit():
        numbers += 1

if letters > 0 and numbers > 0:
    classification = "VALID MIXED SIGNAL"
elif letters > 0:
    classification = "LETTER SIGNAL"
elif numbers > 0:
    classification = "NUMERIC SIGNAL"
else:
    classification = "INVALID SIGNAL"

print("Signal Classification:")
print(classification)
```

```python
# ASSESSMENT DATA

print("=== ASSESSMENT DATA ===")
print(f"Generated Signal: {signal}")
print(f"Processed Signal: {processed_signal}")
print(f"Letters Detected: {letters}")
print(f"Numbers Detected: {numbers}")
print(f"Signal Classification: {classification}")
```

```python
print("Execution Log:")
print("1. Signal generated using engineer name and SEED_NUM.")
print("2. Signal processed and normalized.")
print("3. Signal analyzed character by character.")
print("4. Signal classification completed.")
```

```python
print("=== FINAL OUTPUT ===")
print(f"Engineer: {ENGINEER}")
print(f"Seed Number: {SEED_NUM}")
print(f"Final Signal: {processed_signal}")
print(f"Classification: {classification}")
print("Signal diagnostic completed successfully.")
```

```python
# EXERCISE 3 — AUTHENTICATION

authorized_password = ENGINEER + str(SEED_NUM)
attempt_limit = SEED_NUM

print("Authentication System Initialized")
print(f"Attempt Limit: {attempt_limit}")
```

```python
# AUTHENTICATION PROCESS

attempts = 0
access_granted = False

while attempts < attempt_limit:
    password = input("Enter Password: ")
    attempts += 1

    if password == authorized_password:
        access_granted = True
        print("ACCESS GRANTED")
        break
    else:
        print("ACCESS DENIED")
        print(f"Attempts remaining: {attempt_limit - attempts}")
```

```python
# FINAL SYSTEM STATE

if access_granted:
    system_state = "UNLOCKED"
    access_result = "ACCESS GRANTED"
else:
    system_state = "LOCKED"
    access_result = "ACCESS DENIED"

print(f"Access Result: {access_result}")
print(f"Final System State: {system_state}")
```

```python
# ASSESSMENT DATA

print("=== ASSESSMENT DATA ===")
print(f"Generated Password: {authorized_password}")
print(f"Attempt Limit: {attempt_limit}")
print(f"Attempts Made: {attempts}")
print(f"Access Result: {access_result}")
print(f"Final System State: {system_state}")
```

```python
# FINAL OUTPUT

print("=== FINAL OUTPUT ===")
print(f"Engineer: {ENGINEER}")
print(f"Generated Password: {authorized_password}")
print(f"Attempt Limit: {attempt_limit}")
print(f"Attempts Made: {attempts}")
print(f"Access Result: {access_result}")
print(f"Final System State: {system_state}")
```
